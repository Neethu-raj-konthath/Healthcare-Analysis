```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset

df = pd.read_csv("1776250375-P2-Healthcare Analytics for Doctor Visits (1).csv")

# Display first 5 rows
df.head()
```

```python
plt.figure(figsize=(8,5))

plt.hist(
    df["visits"],
    bins=10,
    edgecolor="black"
)

plt.title("Distribution of Doctor Visits")
plt.xlabel("Number of Doctor Visits")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x="gender"
)

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of People")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.barplot(
    data=df,
    x="gender",
    y="visits"
)

plt.title("Average Doctor Visits by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Doctor Visits")

plt.show()
```

```python
plt.figure(figsize=(9,5))

sns.countplot(
    data=df,
    x="age"
)

plt.title("Age Group Distribution")
plt.xlabel("Age Code")
plt.ylabel("Number of People")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(7,7))

gender_counts = df["gender"].value_counts()

plt.pie(
    gender_counts.values,
    labels=gender_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Gender Distribution")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.boxplot(
    y=df["visits"]
)

plt.title("Distribution of Doctor Visits")
plt.ylabel("Number of Doctor Visits")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="gender",
    y="visits"
)

plt.title("Doctor Visits by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Doctor Visits")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x="gender",
    y="visits"
)

plt.title("Distribution of Doctor Visits by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Doctor Visits")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="private"
)

plt.title("Private Insurance Distribution")
plt.xlabel("Private Insurance")
plt.ylabel("Number of People")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="reduced",
    y="visits"
)

plt.title("Reduced Activity vs Doctor Visits")
plt.xlabel("Reduced Activity")
plt.ylabel("Number of Doctor Visits")

plt.show()
```

```python
plt.figure(figsize=(10,7))

numeric_data = df[
    [
        "visits",
        "age",
        "income",
        "illness",
        "reduced",
        "health"
    ]
]

correlation = numeric_data.corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Healthcare Variables")

plt.show()
```

```python
sns.pairplot(
    df[
        [
            "visits",
            "age",
            "income",
            "illness",
            "reduced",
            "health"
        ]
    ]
)

plt.show()
```

```python
average_visits = df.groupby(
    ["gender", "private"]
)["visits"].mean().reset_index()

plt.figure(figsize=(9,5))

sns.barplot(
    data=average_visits,
    x="gender",
    y="visits",
    hue="private"
)

plt.title("Average Doctor Visits by Gender and Private Insurance")
plt.xlabel("Gender")
plt.ylabel("Average Doctor Visits")

plt.show()
```
